### 2b. Plot the data & estimate initial growth rate (R0) from early data (through day 45)
This section should come from your python code after Data Release #1.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the data
data = pd.read_csv('../Data/mystery_virus_daily_active_counts_RELEASE#1.csv',
                   parse_dates=['date'],
                   header=0,
                   index_col=None)

# Rename long column for easier use
data = data.rename(columns={'active reported daily cases': 'active'})

# Plot active infections over time
plt.figure()
plt.plot(data['date'], data['active'])

plt.xlabel('Date')
plt.ylabel('Active Infections')
plt.title('Mystery Virus: Active Infections Over Time')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# What do you notice about the initial infections?
#The initial infections start very small and increase slowly at first. After a short time, the number of infections begins increasing more rapidly, showing exponential growth in the early phase of the outbreak.
# How could we measure how quickly its spreading?
 #We could measure how quickly it’s spreading by calculating the exponential growth rate (r), the doubling time, or the basic reproduction number (R₀). These values tell us how fast the number of infections is increasing.
# What information about the virus would be helpful in determining the shape of the outbreak curve?
#Important information would include the basic reproduction number (R₀), how long a person is infectious, the incubation period, and how the virus spreads (for example, through air or contact). These factors affect how quickly the outbreak grows and how large it becomes.
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress

# Load the data
data = pd.read_csv('../Data/mystery_virus_daily_active_counts_RELEASE#1.csv',
                   parse_dates=['date'],
                   header=0,
                   index_col=None)

# Rename long column
data = data.rename(columns={'active reported daily cases': 'active'})

# Use early outbreak period for exponential growth estimation
early_data = data.iloc[0:45]

# Convert time to numeric days
t = early_data['day'].values
I = early_data['active'].values

# Take natural log of infections
log_I = np.log(I)

# Fit ln(I) = rt + b
slope, intercept, r_value, p_value, std_err = linregress(t, log_I)

r = slope  # exponential growth rate

print("Estimated growth rate r:", r)



D = 2     # infectious period in days given in powerpoint

R0 = 1 + r * D

print("Estimated R0:", R0)
print("R^2 of exponential fit:", r_value**2)

# Optional: Plot exponential fit
plt.figure()
plt.scatter(t, I, label='Data')
plt.plot(t, np.exp(intercept + r*t), label='Exponential Fit')
plt.xlabel('Day')
plt.ylabel('Active Infections')
plt.legend()
plt.title('Early Exponential Fit')
plt.show()
```
